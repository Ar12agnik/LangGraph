```python
from langgraph.graph import StateGraph,START,END
from langchain_openrouter import ChatOpenRouter
from typing import TypedDict
from dotenv import load_dotenv
```

```python
load_dotenv()
model = model = ChatOpenRouter(
    model="nvidia/nemotron-3-nano-omni-30b-a3b-reasoning:free",
    temperature=0.8,
)
```

```python
class LLMState(TypedDict):
    q:str
    a:str
```

```python
def llmqa(state:LLMState) -> LLMState:
    question = state['q']
    prompt = f"Answer the following question {question} dont include any markdown"
    ans = model.invoke(prompt).content
    state['a']= str(ans)
    return state
```

```python
graph = StateGraph(LLMState)
graph.add_node("llm_qa",llmqa)
graph.add_edge(START,'llm_qa')
graph.add_edge('llm_qa',END)
wf = graph.compile()
```

```python
# initial_state = {"q":"how far is earth to sun measured in moons?","a":""}
fs = wf.invoke({"q":"how far is earth to sun measured in moons?","a":""})
print(fs)
```
